```python
import dask.bag as db
```

```python
bag = db.read_text('wse stocks\\*.txt')
```

```python
def count_lines(partition):
    return [len(partition)]

lines_per_file = bag.map_partitions(count_lines)
```

```python
lines_per_file_list = lines_per_file.compute()
```

```python
files_with_2_or_more_lines = sum(1 for x in lines_per_file_list if x >= 2)

print("Liczba plików z co najmniej 2 liniami:", files_with_2_or_more_lines)
```
